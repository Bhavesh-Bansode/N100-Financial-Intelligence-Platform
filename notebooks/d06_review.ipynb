```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/nifty100.db")
```

```python
query = """
SELECT id
FROM companies
ORDER BY RANDOM()
LIMIT 5
"""

companies = pd.read_sql(query, conn)

companies
```

```python
tables = [
    "profitandloss",
    "balancesheet",
    "cashflow",
    "stock_prices",
    "market_cap"
]

tickers = [
    "SBIN",
    "PIDILITIND",
    "KOTAKBANK",
    "TATAPOWER",
    "BPCL"
]

for ticker in tickers:

    print("\n" + "="*60)
    print(ticker)
    print("="*60)

    for table in tables:

        query = f"""
        SELECT COUNT(*) AS row_count
        FROM {table}
        WHERE company_id = '{ticker}'
        """

        result = pd.read_sql(query, conn)

        print(
            f"{table}:",
            result.iloc[0]["row_count"]
        )
```

```python
for ticker in [
    "PIDILITIND",
    "KOTAKBANK",
    "TATAPOWER",
    "BPCL"
]:

    print("\n", ticker)

    for table in [
        "profitandloss",
        "balancesheet",
        "cashflow",
        "stock_prices",
        "market_cap"
    ]:

        q = f"""
        SELECT COUNT(*) as cnt
        FROM {table}
        WHERE company_id = '{ticker}'
        """

        cnt = pd.read_sql(q, conn)["cnt"][0]

        print(table, cnt)
```
