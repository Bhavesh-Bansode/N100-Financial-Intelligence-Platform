```python
check = pd.read_sql("""
SELECT
    fr.company_id,
    fr.year,
    ROUND(fr.return_on_equity_pct, 2) AS calculated_roe,
    c.roe_percentage AS source_roe
FROM financial_ratios fr
JOIN companies c
ON fr.company_id = c.id
LIMIT 20;
""", conn)

print(check)
```

```python
latest_roe = pd.read_sql("""
SELECT
    fr.company_id,
    fr.year,
    ROUND(fr.return_on_equity_pct, 2) AS calculated_roe,
    c.roe_percentage AS source_roe
FROM financial_ratios fr
JOIN companies c
    ON fr.company_id = c.id
WHERE fr.year = '2024-03'
ORDER BY fr.company_id;
""", conn)

print(latest_roe.head(20))
```
